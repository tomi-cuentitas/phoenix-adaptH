```python


import math
import numpy as np
import pytest

from phoenix.keymap import KeyMap, Entry
from phoenix.adaa_derived import FortranRA, NumPyRA

from adaptHeisenberg.sigma0 import ProductState
from adaptHeisenberg.covariance_scalar_product import fetch_covar_scalar_product

def _build_system_keymap(keys):
    root = KeyMap(name="sys")
    
    by_nums = {}
    for nums, word in keys: 
        by_nums.setdefault(nums, set()).add(word)
        
    for nums, words in by_nums.items():
        sub = KeyMap(name = f"sub_{nums}")
        sub.labels = (nums,)
        
        for w in sorted(words):
            sub.link(w, Entry(name = f"w = {w}"))
        root.link(nums, sub)
    
    root.update()
    
    return root
    

def test_covar_sp_results_from_pj_id():
    
    ### first we start with short strings , just to 
    ### check all possible values for tj 
    
    
    sigma0_z = ProductState([.25, .75, 1., 
                             .25, .75, 1.])
    sp_local = fetch_covar_scalar_product(sigma0_z)
    
    p_src_terms = {
        ((), ""): 1.,
        ((0,), "x"): 1.,
        ((0,), "y"): 2.,
        ((0,), "z"): 3.,
    }
    
    all_keys = set(p_src_terms.keys())
    km = _build_system_keymap(all_keys)
    op = FortranRA.set_keymap(km)
    
    tgt = op().to_zero()
    off_id = km.key2off((), "")
    tgt.data["real"][off_id] = 1.0
    
    for key, coeff in p_src_terms.items():       
        src = op()
        src.to_zero()
        off = km.key2off(*key)
        src.data["real"][off] = float(coeff)

        if key == ((), ""):
            expected = coeff
        else:
            axis = {"x": 0, "y": 1, "z": 2}[key[1]]
            expected = coeff * sigma0_z.vectors[0, axis]
            
        
        assert sp_local(src, tgt) == expected
        
    for key, coeff in p_src_terms.items():       
        src = op()
        src.to_zero()
        off = km.key2off(*key)
        src.data["real"][off] = float(coeff)

        if key == ((), ""):
            expected = coeff
        else:
            axis = {"x": 0, "y": 1, "z": 2}[key[1]]
            expected = coeff * sigma0_z.vectors[0, axis]
            
        
        assert sp_local(tgt, src) == expected
        
    for key, coeff in p_src_terms.items():       
        src = op()
        src.to_zero()
        off = km.key2off(*key)
        src.data["real"][off] = float(coeff)

        if key == ((), ""):
            expected = coeff
        else:
            axis = {"x": 0, "y": 1, "z": 2}[key[1]]
            expected = coeff * sigma0_z.vectors[0, axis]
            
        
        assert sp_local(src, src) == expected
    
    
def test_covar_sp_results_from_pj_qj():
    
    sigma0_z = ProductState([.25, .75, 1., 
                             .25, .75, 1.])
    sp_local = fetch_covar_scalar_product(sigma0_z)
    
    p_src_terms = {
        ((0,1), "xx"): 0.1,
        ((0,1), "xy"): -0.2,
        ((0,), "xz"): 0.3,
        ((0,1), "yx"): 1.,
        ((0,1), "yy"): 2.,
        ((0,1), "yz"): 3.,
        ((0,1), "zx"): 1.,
        ((0,1), "zy"): 2.,
        ((0,1), "zz"): 3.,
        ((0), ""): 1,
    }

```
